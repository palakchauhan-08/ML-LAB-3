```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Karan", "Sneha", "Vikas", "Neha"],
    "Age": [20, 21, None, 22, 21, 20, 23, None],
    "Gender": ["Male", "Female", None, "Female", "Male", "Female", "Male", None],
    "Marks": [85, None, 78, 88, 67, 95, None, 81],
    "Department": ["CSE", "IT", "CSE", "ECE", None, "CSE", "ECE", "IT"]
}

df = pd.DataFrame(data)

df.to_csv("student_data_missing.csv", index=False)

print("Dataset created successfully!")
print(df)
```

```python
import pandas as pd

df = pd.read_csv("student_data_missing.csv")

print(df.isnull().sum())
```

```python
print(df.isnull().mean() * 100)
```

```python
df_clean = df.dropna()

print(df_clean)
```

```python
df_clean = df.dropna(axis=1, thresh=len(df) * 0.5)

print(df_clean)
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

print(df)
```

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

print(df)
```

```python
df = df.ffill()

print(df)
```

```python
df = df.bfill()

print(df)
```

```python
before = pd.read_csv("student_data_missing.csv")

after = before.dropna()

print("Before:")
print(before)

print("\nAfter:")
print(after)
```

```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Marks"] < lower) | (df["Marks"] > upper)]

print(outliers)
```

```python
from scipy.stats import zscore

df["Z_score"] = zscore(df["Marks"])

outliers = df[abs(df["Z_score"]) > 3]

print(outliers)
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Marks"].dropna())

plt.title("Box Plot of Marks")
plt.ylabel("Marks")
plt.show()
```

```python
plt.scatter(df["Student_ID"], df["Marks"])

plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("Scatter Plot")
plt.show()
```

```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_iqr = df[(df["Marks"] >= lower) & (df["Marks"] <= upper)]

print(df_iqr)
```

```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

median = df["Marks"].median()

df["Marks"] = df["Marks"].apply(
    lambda x: median if x < lower or x > upper else x
)

print(df)
```

```python
lower = df["Marks"].quantile(0.05)
upper = df["Marks"].quantile(0.95)

df["Marks"] = df["Marks"].clip(lower, upper)

print(df)
```

```python
before = pd.read_csv("student_data_missing.csv")

Q1 = before["Marks"].quantile(0.25)
Q3 = before["Marks"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

after = before[
    (before["Marks"] >= lower) &
    (before["Marks"] <= upper)
]

print("Before:")
print(before)

print("\nAfter:")
print(after)
```
